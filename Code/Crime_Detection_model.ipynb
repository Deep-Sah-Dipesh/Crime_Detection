```python
import os
import torch
import numpy as np
import logging
import sys
from tqdm import tqdm
import time # For get_timestamp if used in logger setup

# --- Setup Project Path & Imports ---
# Adjust this if your config.py is not in the parent directory of your notebook
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.append(project_root)

# Default/Fallback configuration in case config.py cannot be loaded or is incomplete
class DefaultConfig:
    BASE_PATHS = {
        'Training': {
            'Videos': r'G:\Crime_Detection\Training\Videos',
            'Processed_Data': r'G:\Crime_Detection\Training\Processed_Data' # Example path, correct this to your actual path
        }
    }
    OUTPUT_DIRS = {
        'checkpoints': r'G:\Crime_Detection\checkpoints', # Example path, correct this to your actual path
        'logs': r'G:\Crime_Detection\logs' # Example path, correct this to your actual path
    }
    RANDOM_SEED = 42

# Attempt to import from config.py, fall back to DefaultConfig if issues arise
try:
    from config import BASE_PATHS as Imported_BASE_PATHS, OUTPUT_DIRS as Imported_OUTPUT_DIRS, RANDOM_SEED as Imported_RANDOM_SEED
    BASE_PATHS = Imported_BASE_PATHS
    OUTPUT_DIRS = Imported_OUTPUT_DIRS
    RANDOM_SEED = Imported_RANDOM_SEED
    logger = logging.getLogger('notebook_feature_repair') # Initialize logger here to use it in error handling

    # Explicitly check for the 'Training' key structure
    if 'Training' not in BASE_PATHS or \
       'Videos' not in BASE_PATHS.get('Training', {}) or \
       'Processed_Data' not in BASE_PATHS.get('Training', {}):
        logger.error("Error: BASE_PATHS in config.py does not contain the expected 'Training' structure ('Videos' and 'Processed_Data').")
        logger.error("Falling back to internal default paths. Please correct your config.py.")
        BASE_PATHS = DefaultConfig.BASE_PATHS
        OUTPUT_DIRS = DefaultConfig.OUTPUT_DIRS
        RANDOM_SEED = DefaultConfig.RANDOM_SEED
    else:
        logger.info("Successfully loaded configuration from config.py.")

except ImportError as e:
    BASE_PATHS = DefaultConfig.BASE_PATHS
    OUTPUT_DIRS = DefaultConfig.OUTPUT_DIRS
    RANDOM_SEED = DefaultConfig.RANDOM_SEED
    logger = logging.getLogger('notebook_feature_repair') # Initialize logger for fallback case
    logger.error(f"Error importing config.py: {e}. Falling back to internal default paths. Please ensure config.py is in your project root.")

# --- Logger Setup for Notebook ---
if not logger.handlers:
    logger.setLevel(logging.INFO)
    handler = logging.StreamHandler(sys.stdout)
    formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
    handler.setFormatter(formatter)
    logger.addHandler(handler)
    logger.propagate = False
# --- End Logger Setup ---

# --- Feature Repair & Extraction Functions ---

def get_clip_id_from_path(video_path, base_video_dir):
    relative_path = os.path.relpath(video_path, base_video_dir)
    return os.path.splitext(relative_path)[0].replace(os.sep, '__')

def get_expected_feature_paths(clip_id, base_processed_data_dir):
    # Generates expected feature paths. Adjust this if your data structure differs.
    range_folder = ""
    if "Kill.Bill" in clip_id or "Mission.Impossible" in clip_id or "Mindhunters" in clip_id or "Haywire" in clip_id:
        range_folder = "1-1004"
    elif "Skyfall" in clip_id:
        range_folder = "1005-2004"
    elif "ekH2zovSEIE" in clip_id:
        range_folder = "2005-2804"

    audio_path = os.path.join(base_processed_data_dir, 'audio_features', range_folder, clip_id, f"{clip_id}_yamnet_embeddings.npy")
    video_path = os.path.join(base_processed_data_dir, 'video_features', range_folder, clip_id, f"{clip_id}_all.npy")
    text_path = os.path.join(base_processed_data_dir, 'text_features', range_folder, f"{clip_id}_roberta_features.npy")
    
    return {
        'audio': audio_path,
        'video': video_path,
        'text': text_path
    }

def extract_audio_feature_for_clip(original_video_path, output_feature_path, current_logger):
    # REPLACE THIS WITH YOUR ACTUAL YAMNET AUDIO FEATURE EXTRACTION LOGIC.
    current_logger.info(f"Re-extracting audio for {os.path.basename(original_video_path)}")
    try:
        os.makedirs(os.path.dirname(output_feature_path), exist_ok=True)
        # Your code here (e.g., torchaudio.load, YamnetModel, np.save)
        dummy_data = np.random.rand(10, 1024)
        np.save(output_feature_path, dummy_data)
        current_logger.info(f"Successfully re-extracted audio: {os.path.basename(output_feature_path)}")
        return True
    except Exception as e:
        current_logger.error(f"Failed to re-extract audio for {os.path.basename(original_video_path)}: {e}")
        return False

def extract_video_feature_for_clip(original_video_path, output_feature_path, current_logger):
    # REPLACE THIS WITH YOUR ACTUAL VIDEO FEATURE EXTRACTION LOGIC.
    current_logger.info(f"Re-extracting video for {os.path.basename(original_video_path)}")
    try:
        os.makedirs(os.path.dirname(output_feature_path), exist_ok=True)
        # Your code here (e.g., cv2.VideoCapture, video model, np.save)
        dummy_data = np.random.rand(20, 512)
        np.save(output_feature_path, dummy_data)
        current_logger.info(f"Successfully re-extracted video: {os.path.basename(output_feature_path)}")
        return True
    except Exception as e:
        current_logger.error(f"Failed to re-extract video for {os.path.basename(original_video_path)}: {e}")
        return False

def extract_text_feature_for_clip(original_video_path, output_feature_path, current_logger):
    # REPLACE THIS WITH YOUR ACTUAL ROBERTA TEXT FEATURE EXTRACTION LOGIC.
    # This likely involves reading a transcript file associated with the video.
    current_logger.info(f"Re-extracting text for {os.path.basename(original_video_path)}")
    try:
        os.makedirs(os.path.dirname(output_feature_path), exist_ok=True)
        # Your code here (e.g., transformers, tokenizer, model, np.save)
        dummy_data = np.random.rand(768)
        np.save(output_feature_path, dummy_data)
        current_logger.info(f"Successfully re-extracted text: {os.path.basename(output_feature_path)}")
        return True
    except Exception as e:
        current_logger.error(f"Failed to re-extract text for {os.path.basename(original_video_path)}: {e}")
        return False

def scan_video_directory(video_dir):
    # Scans video directory for all video files.
    video_files = []
    for root, _, files in os.walk(video_dir):
        for file in files:
            if file.lower().endswith(('.mp4', '.avi', '.mov', '.mkv')):
                video_files.append(os.path.join(root, file))
    return video_files

def run_feature_repair_in_notebook():
    logger.info("Starting missing feature repair process in notebook...")
    
    video_source_dir = BASE_PATHS['Training']['Videos']
    processed_data_root = BASE_PATHS['Training']['Processed_Data']
    
    os.makedirs(processed_data_root, exist_ok=True)
    os.makedirs(os.path.join(processed_data_root, 'audio_features'), exist_ok=True)
    os.makedirs(os.path.join(processed_data_root, 'video_features'), exist_ok=True)
    os.makedirs(os.path.join(processed_data_root, 'text_features'), exist_ok=True)

    all_video_paths = scan_video_directory(video_source_dir)
    
    total_clips = len(all_video_paths)
    missing_features_count = 0
    repaired_features_count = 0

    if total_clips == 0:
        logger.warning("No video files found in source directory. Cannot check for features.")
        return

    logger.info(f"Found {total_clips} video clips to check for features.")

    for i, video_path in enumerate(tqdm(all_video_paths, desc="Checking/Repairing Features")):
        clip_id = get_clip_id_from_path(video_path, video_source_dir)
        expected_paths = get_expected_feature_paths(clip_id, processed_data_root)

        for modality, feature_path in expected_paths.items():
            if not os.path.exists(feature_path):
                missing_features_count += 1
                if modality == 'audio':
                    if extract_audio_feature_for_clip(video_path, feature_path, logger):
                        repaired_features_count += 1
                elif modality == 'video':
                    if extract_video_feature_for_clip(video_path, feature_path, logger):
                        repaired_features_count += 1
                elif modality == 'text':
                    if extract_text_feature_for_clip(video_path, feature_path, logger):
                        repaired_features_count += 1
                else:
                    logger.error(f"Unknown modality '{modality}' for repair.")
    
    logger.info("Feature repair process completed.")
    logger.info(f"Total features found missing: {missing_features_count}")
    logger.info(f"Total features re-extracted: {repaired_features_count}")
    if missing_features_count > 0 and repaired_features_count < missing_features_count:
        logger.warning("Some features were found missing but could not be repaired. Check previous error logs for details.")

# --- Execute the repair process ---
run_feature_repair_in_notebook()

print("\nFeature repair process finished. You can now proceed with your training script or dataset loading in this notebook.")
```
